# NADI Canonical Evidence Notebook

**Current canonical result.** NADI is a retrospective mixed-environmental-water testing-prioritization system. This notebook keeps laboratory, hydraulic, and simulated evidence separate.

## 1. Project scope and claim boundary
The project does not demonstrate direct E. coli sensing, drinking-water pipe localization, Delhi-NCR deployment, avoided illness, or an integrated field event.

## 2. Evidence-track map

| Track | Evidence label | Valid output |
|---|---|---|
| EA E. coli | AUTHENTIC / LAB_CONFIRMED | retrospective risk and testing priority |
| BWFLnet | AUTHENTIC / OBSERVED pressure | pressure reconstruction |
| Net3/L-TOWN | SIMULATED | conditional decision stress |
| GECCO/Texas/prior adaptation | Separate supporting track | domain-specific stress evidence |

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent

auth_dir = ROOT / 'results/canonical/authentic/audit_20261120_07'
hydro_dir = ROOT / 'results/canonical/hydraulics/audit_20261120_01'
sim_dir = ROOT / 'results/canonical/simulation/audit_20261120_01'
auth = json.loads((auth_dir / 'authentic_experiment_report.json').read_text())
hydro = json.loads((hydro_dir / 'hydraulic_evaluation.json').read_text())
simulation = json.loads((sim_dir / 'net3_simulation_summary.json').read_text())
summary = pd.read_csv(auth_dir / 'authentic_test_summary.csv').set_index('policy')
print('AUTHENTIC:', auth['data_policy']['domain'])
print('HYDRAULIC:', hydro['domain'])
print('SIMULATION:', simulation['domain'])
```

## 3. Authentic dataset inventory
The processed EA table has 13,405 unique site-day records, 920 sites, and 285 dates. Its raw materials include sea, river, estuarine, drainage, and other environmental-water observations.

## 4. Chronological causal feature construction
Each date receives site, region, and global historical features before any outcome from that date updates history. `feature_cutoff` is strictly before `round_date`.

## 5. Model-family comparison
Logistic, XGBoost, and histogram-gradient model families are trained through May 2024. Model selection occurs on the validation period, before locked evaluation.

## 6. Calibration
Platt calibration and threshold selection use June–July only. The calibration bundle is hash-pinned by the locked evaluator.

```python
best = auth['best_model_selected_on_validation']
locked = auth['model_reports'][best]['locked_test']
print({'model': best, 'n': locked['n'], 'F1': round(locked['f1'], 3), 'ROC_AUC': round(locked['roc_auc'], 3), 'PR_AUC': round(locked['pr_auc'], 3), 'sensitivity': round(locked['sensitivity'], 3), 'specificity': round(locked['specificity'], 3)})
assert locked['n'] == 545
assert locked['specificity'] < 0.5  # retained limitation, not hidden
```

## 7. SDAIF mathematical formulation
For a finite uncertainty set, SDAIF selects a fixed-budget plan minimizing posterior/robust loss and reports worst-case regret. It carries calibrated-model disagreement into test selection.

## 8. Fixed-budget policy comparison
All authentic policies receive the same observed candidate set and three-test budget.

## 9. Sequential update demonstration
The stored replay updates after a selected historical outcome. This is an analytical same-round reveal assumption; real laboratory turnaround is not observed.

## 10. Regret and adversarial-model explanation
Decision records retain the model-disagreement and worst-case regret quantities used by the selector.

```python
print(summary[['rounds', 'total_positives_found', 'mean_positive_discovery_regret']].to_string())
assert summary.loc['adaptive_sdaif', 'total_positives_found'] == 114
assert summary.loc['adaptive_sdaif', 'total_positives_found'] == summary.loc['adaptive_bayesian', 'total_positives_found']
assert summary.loc['adaptive_sdaif', 'total_positives_found'] == summary.loc['best_single_model', 'total_positives_found']
# No superiority claim follows from this tie.
```
